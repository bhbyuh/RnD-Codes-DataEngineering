```sql
create schema warehousefactdim;
```

```sql
-- staging/raw/bronze
CREATE or REPLACE TABLE warehousefactdim.stg_sales
AS
SELECT * FROM warehousesourcefactdim.Orders
```

```sql
-- now will create view for that transofmration to make it for core layer 
CREATE VIEW warehousefactdim.trans_sales
AS
SELECT * FROM warehousefactdim.stg_sales where Quantity IS NOT NULL
```

```sql
-- core layer 
SELECT * FROM warehousefactdim.trans_sales 
```

```sql
-- -- Dimension Customer 
CREATE TABLE warehousefactdim.DimCustomer(
    DimCustomerKey INT,
    CustomerID INT,
    CustomerName STRING,
    CustomerEmail STRING
);

-- Now we will create fact and dimension tables. Fact and dimension tables typically follow a many to one relationship.
-- We will create the dimension tables first.
-- Dimension Customer 
INSERT INTO warehousefactdim.DimCustomer
SELECT row_number() OVER (ORDER BY T.CustomerID) as DimCustomerKey,T.* FROM -- DimCustomerKey is a surrogate key 
(
    SELECT  DISTINCT(CustomerID) as CustomerID,
        CustomerName,
        CustomerEmail 
    FROM warehousefactdim.trans_sales
) AS T;


SELECT * FROM warehousefactdim.DimCustomer
```

```sql
-- Dimension Product 
CREATE TABLE warehousefactdim.dimproduct(
    DimprodKey INT,
    prodID INT,
    prodName STRING,
    prodcategory STRING
);

INSERT INTO warehousefactdim.dimproduct
SELECT row_number() OVER (ORDER BY T.productID) as DimprodKey,T.* FROM -- DimprodKey is a surrogate key 
(
    SELECT  DISTINCT(ProductID) as productID,
        ProductName,
        ProductCategory 
    FROM warehousefactdim.trans_sales
) AS T;


SELECT * FROM warehousefactdim.dimproduct
```

```sql
-- Dimension Region 
CREATE TABLE warehousefactdim.dim_region(
    DimregionKey INT,
    regionID INT,
    regionName STRING,
    Country STRING
);

INSERT INTO warehousefactdim.dim_region
SELECT row_number() OVER (ORDER BY T.regionID) as DimregionKey,T.* FROM -- DimregionKey is a surrogate key 
(
    SELECT  DISTINCT(RegionID) as regionID,
        RegionName,
        Country 
    FROM warehousefactdim.trans_sales
) AS T;

SELECT * FROM warehousefactdim.dim_region
```

```sql
-- dim date 

CREATE TABLE warehousefactdim.dim_date(
    DimDateKey INT,
    orderDate DATE
);

INSERT INTO warehousefactdim.dim_date
SELECT row_number() OVER (ORDER BY T.OrderDate) as DimdateKey,T.* FROM -- DimdateKey is a surrogate key 
(
    SELECT  DISTINCT(OrderDate) as OrderDate
    FROM warehousefactdim.trans_sales
) AS T;

SELECT * FROM warehousefactdim.dim_date
```

```sql
-- Fact table
CREATE TABLE warehousefactdim.factsales(
        OrderID INT, 
        Quantity DECIMAL,
        UnitPrice DECIMAL,
        TotalAmount DECIMAL,
        DimCustomerKey INT,
        DimprodKey INT,
        DimregionKey INT,
        DimDateKey INT
)
```

```sql
SELECT 
    F.OrderID,
    F.Quantity,
    F.UnitPrice,
    F.TotalAmount,
    DC.DimCustomerKey,
    DP.DimprodKey,
    DR.DimregionKey,
    DD.DimDateKey
FROM
    warehousefactdim.trans_sales F
LEFT JOIN warehousefactdim.DimCustomer DC ON 
    F.CustomerID = DC.CustomerID
LEFT JOIN warehousefactdim.dimproduct DP ON 
    F.ProductID = DP.prodID
LEFT JOIN warehousefactdim.dim_region DR ON 
    F.Country = DR.Country -- do see why country here , concept of granularity  
LEFT JOIN warehousefactdim.dim_date DD ON 
    F.OrderDate = DD.orderDate

```

```sql
create table democatalog.warehouse.scdtype2_src(
    productid int,
    productname string,
    productcategory string,
    processDate date
)
```

```sql
create table democatalog.warehouse.scdtype2_table(
    productid int,
    productname string,
    productcategory string,
    processDate date,
    start_date date,
    end_date date,
    isactive boolean
) 
```

```sql
CREATE TEMPORARY VIEW scdview AS
SELECT * FROM democatalog.warehouse.scdtype2_src
```

```sql
-- this is just to update previous record set end date and isactive = false

MERGE INTO democatalog.warehouse.scdtype2_table trg
USING scdview
ON trg.productid = scdview.productid AND trg.isactive = true
WHEN MATCHED AND 
 ( trg.productcategory <> scdview.productcategory OR  -- <> like not equal 
 trg.productname <> scdview.productname OR
 trg.productcategory <> scdview.productcategory ) THEN UPDATE SET trg.end_date=current_timestamp() , trg.isactive = false
```

```sql
-- to insert new record

MERGE INTO democatalog.warehouse.scdtype2_table trg
USING scdview
ON trg.productid = scdview.productid AND trg.isactive = true
WHEN NOT MATCHED THEN INSERT (productid,productname,productcategory,processDate,start_date,end_date,isactive) 
VALUES (scdview.productid,scdview.productname,scdview.productcategory,scdview.processDate,current_timestamp(),'9999-12-31',true)
```
